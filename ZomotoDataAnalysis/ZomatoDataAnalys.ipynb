```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# Create The Data Frame

```python
dataframe =pd.read_csv("Zomato data .csv")
dataframe.head()
```

# Covert the Data Type of column Rate 

```python
def handleRate(value):
    value = str(value).split('/')
    value=value[0];
    return float(value)
dataframe['rate']=dataframe['rate'].apply(handleRate)
dataframe.head()
```

```python
dataframe.info()
```

# Type Of Resturant

```python
dataframe.head()
```

```python
sns.countplot(x=dataframe['listed_in(type)'])
plt.xlabel('type of resturant')
```

# conclusion - majority of the resturant falls in dining category

```python
dataframe.head()
```

```python
grouped_data=dataframe.groupby('listed_in(type)')['votes'].sum()
result = pd.DataFrame({'votes':grouped_data})
plt.plot(result, c="green",marker="o")
plt.xlabel("Type of resturant ",c="red",size=20)
plt.ylabel("votes ",c="red",size=20)
```

# conclusion- Dinning resturants has recieved maximum votes

```python
dataframe.head()
```

```python
#dataframe['rate'].value_counts()
plt.hist(dataframe['rate'],bins=10)
plt.title("rating disrtibution")
plt.show()
```

# conclusion - the majority resturant received majority rating from 3.5 to 4 (3.8)
